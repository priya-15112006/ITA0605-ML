```python
9. Linear and Polynomial Regression Comparison
Aim: To compare Linear Regression and Polynomial Regression using Python.

Procedure:
1. Import the required libraries.
2. Load the dataset.
3. Prepare the input and output variables.
4. Create a Linear Regression model.
5. Create a Polynomial Regression model.
6. Train both models using the dataset.
7. Compare their performance using evaluation metrics and graphs.
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Training data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 5, 10, 17, 26])

# Linear Regression
linear = LinearRegression()
linear.fit(X, y)
linear_pred = linear.predict(X)

# Polynomial Regression
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)
poly_pred = model.predict(X_poly)

# Display results
print("Actual Values:", y)
print("Linear Predictions:", linear_pred)
print("Polynomial Predictions:", poly_pred)
```

```python
10. Expectation-Maximization Algorithm
Aim: To implement the Expectation-Maximization (EM) algorithm using Python.

Procedure:
1. Import the required libraries.
2. Load the dataset.
3. Initialize the model parameters.
4. Perform the Expectation step to calculate probabilities.
5. Perform the Maximization step to update parameters.
6. Repeat both steps until convergence.
7. Display the final results.
import numpy as np
from sklearn.mixture import GaussianMixture

# Sample data
X = np.array([[1], [2], [3], [8], [9], [10]])

# Create EM model
model = GaussianMixture(n_components=2, random_state=0)

# Train the model
model.fit(X)

# Predict clusters
prediction = model.predict(X)

print("Cluster Labels:", prediction)
print("Cluster Means:", model.means_)
```

```python
12. Iris Flower Classification using KNN
Aim: To classify Iris flowers into different species using the K-Nearest Neighbours algorithm.

Procedure:
1. Import the required libraries.
2. Load the Iris dataset.
3. Separate input features and target labels.
4. Split the dataset into training and testing sets.
5. Create the KNN classifier.
6. Train the model and predict flower species.
7. Display the classification accuracy.
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier

# Load Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# Create KNN model
model = KNeighborsClassifier(n_neighbors=3)

# Train the model
model.fit(X, y)

# Predict flower type
prediction = model.predict([[5.1, 3.5, 1.4, 0.2]])

print("Predicted Flower:", iris.target_names[prediction[0]])
```
